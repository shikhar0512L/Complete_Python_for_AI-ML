````python
# ## Difference between `for` and `while` loops in Python

# ### `For` Loop

# *   **Purpose**: Used for iterating over a sequence (like a list, tuple, dictionary, string, or range) or other iterable objects.
# *   **When to use**: Ideal when you know in advance how many times you want to loop, or when you need to process each item in a collection.
# *   **Structure**: Typically consists of an initialization, a condition for iteration, and an automatic increment/decrement (or advancement to the next item in the sequence).
# *   **Example**:
#     ```python
#     for i in range(5):
#         print(i) # Prints 0, 1, 2, 3, 4
#     ```

# ### `While` Loop

# *   **Purpose**: Used for repeatedly executing a block of code as long as a certain condition is true.
# *   **When to use**: Ideal when you don't know the exact number of iterations beforehand, and the loop needs to continue until a specific condition is met (e.g., waiting for user input, processing data until a certain value is reached).
# *   **Structure**: Consists of a condition, and the iteration depends entirely on that condition remaining `True`. You must manually manage any variables that affect the condition within the loop to avoid infinite loops.
# *   **Example**:
#     ```python
#     count = 0
#     while count < 5:
#         print(count)
#         count += 1 # Prints 0, 1, 2, 3, 4
#     ```

# ### Key Differences Summary

# | Feature        | `For` Loop                               | `While` Loop                                      |
# | :------------- | :--------------------------------------- | :------------------------------------------------ |
# | **Usage**      | Iterating over sequences/iterables       | Repeating as long as a condition is true          |
# | **Iterations** | Known number of iterations (often fixed) | Unknown number of iterations (condition-dependent)|
# | **Control**    | Automatically manages iteration progress | Manual management of iteration variable           |
# | **Best for**   | Iterating through collections            | Repeating until a specific event occurs           |
````

````python
class Bags:
  def __init__(self,material,zips,pockets): # self -> it captures the location of variable
    self.material=material
    self.zips=zips
    self.pockets=pockets
    print("This is a constructor. It runs everytime when the class is called")
  name="Gucci"

  def details(self):
    print("Hello this is the company which creates bag. ")

obj = Bags("leather",3,2)
# explaination -> let python created obj at 100 location. So self will store 100, then self.material=material will store 100.material=leather and similarly 100.zips=3, 100.pockets=2
obj2=Bags("polyster",2,4)
print(obj2)
print(obj.name)
obj.details()
````

````python
# Class attribute -> A normal variable created inside a class
# Instance attribute -> attribute created using instances like self.name, aself.age etc..

class Car:
  wheels=4 # class attribute
  def __init__(self,name,color):
    self.name = name
    self.color = color # instance attribute

  def drive(self): # Instance/object method -> captures the location of object
    print(f"{self.name} is driving")

  def stop(self):
    print(f"{self.name} is stopped")

  @classmethod
  def wheelsX(cls):
    print(f"There are {cls.wheels} number of wheels here")

  @staticmethod
  def static_method(): # this is a static method and it will not target any location
    print("This is a static method")

car1=Car("lambo","black")
print(car1.color)
print(car1.name)
print(car1.wheelsX())
car1.drive()
car1.stop()
````

````python
# Inheritance -> allows a child class (or derived class) to inherit attributes and methods from a parent class (or base class). This promotes code reusability, making your programs easier to write, maintain, and scale.
````

````python
class Animal: # Parent class
    def breathe(self):
        print("Breathing...")

class Dog(Animal):     # Dog inherits from Animal
    def bark(self):
        print("Woof!")

class Zebra(Animal):
    def run(self):
        print("Running...")

d = Dog()
d.breathe()   # inherited from Animal ✅
d.bark()      # Dog's own method ✅

z = Zebra()
z.breathe()   # inherited from Animal ✅
z.run()       # Zebra's own method ✅
````

````python
class Parents:
  def __init__(self,hair_color,eye_color,height):
    self.hair_color=hair_color
    self.eye_color=eye_color
    self.height=height

  def display1(self):
    print(f"Inherited from parents: Hair color: {self.hair_color}, Eye color: {self.eye_color}, Height: {self.height}")


class Child(Parents):
  def alive(self):
    print("I am the child")

ch=Child("Black","Brown",5.11)
ch.display1()
ch.alive()

# Your child class objects has all the powers to access the attributes and methods of parent class
````

````python
class BagFactory:
  def __init__(self,material,zips,pockets):
    self.material=material
    self.zips=zips
    self.pockets=pockets

  def details(self):
    print("Your bag details are: ")
    print(self.material)
    print(self.zips)
    print(self.pockets)

class Reebok(BagFactory):
  def __init__(self,material,zips,pockets,color):
    super().__init__(material,zips,pockets) #super() lets a child class call a method from its parent class.
    self.color=color
  def details(self):
    print(self.color)
    return super().details()
  print("This is Reebok class")

bag1=BagFactory("leather",3,4)
bag2=Reebok("polyster",4,2,"black")
print(bag1.details())
print(bag2.details())
````

````python
# Types of Inheritance ->
# 1️⃣ SINGLE INHERITANCE
# One child class inherits from ONE parent class.

# Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name}, {self.age} years old.")

# Child Class (inherits from Person only)
class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)  # Call parent constructor
        self.student_id = student_id

    def show_student_id(self):
        print(f"My Student ID is: {self.student_id}")

    def study(self):
        print(f"{self.name} is studying for exams! 📚")

# Usage
student1 = Student("Shikhar", 18, "S12345")
student1.introduce()        # ✅ Inherited from Person
student1.show_student_id()  # ✅ Student's own method
student1.study()            # ✅ Student's own method
````

````python
# 2️⃣ MULTIPLE INHERITANCE -> One child class inherits from MULTIPLE parent classes.
# Parent 1
class Academic:
    def __init__(self, department, semester):
        self.department = department
        self.semester = semester

    def show_academic_info(self):
        print(f"Department: {self.department}")
        print(f"Semester: {self.semester}")

# Parent 2
class Sports:
    def __init__(self, sport, level):
        self.sport = sport
        self.level = level

    def show_sports_info(self):
        print(f"Sport: {self.sport}")
        print(f"Level: {self.level}")

# Parent 3
class Hostel:
    def __init__(self, hostel_name, room_no):
        self.hostel_name = hostel_name
        self.room_no = room_no

    def show_hostel_info(self):
        print(f"Hostel: {self.hostel_name}")
        print(f"Room No: {self.room_no}")

# Child (inherits from ALL three)
class UniversityStudent(Academic, Sports, Hostel):
    def __init__(self, name, department, semester, sport, level, hostel_name, room_no):
        self.name = name
        Academic.__init__(self, department, semester)
        Sports.__init__(self, sport, level)
        Hostel.__init__(self, hostel_name, room_no)

    def full_profile(self):
        print(f"📋 STUDENT PROFILE: {self.name}")
        print("-" * 30)
        self.show_academic_info()
        self.show_sports_info()
        self.show_hostel_info()

# Usage
# here student can access the method of Academic, Sports, Hostel
student = UniversityStudent(
    "Shikhar singh",
    "Computer Science",
    "3rd",
    "Badminton",
    "University Level",
    "KP-7 AB",
    "3B-445"
)
student.full_profile()
````

````python
# 3️⃣ MULTILEVEL INHERITANCE -> A chain of inheritance: Grandparent → Parent → Child
# Grandparent
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name}")

# Parent (inherits from Person)
class Employee(Person):
    def __init__(self, name, age, employee_id, salary):
        super().__init__(name, age)
        self.employee_id = employee_id
        self.salary = salary

    def show_employee_info(self):
        print(f"Employee ID: {self.employee_id}")
        print(f"Salary: ₹{self.salary:,}")

# Child (inherits from Employee)
class Professor(Employee):
    def __init__(self, name, age, employee_id, salary, subject, research_area):
        super().__init__(name, age, employee_id, salary)
        self.subject = subject
        self.research_area = research_area

    def show_professor_info(self):
        print(f"Subject: {self.subject}")
        print(f"Research Area: {self.research_area}")

    def teach(self):
        print(f"👨‍🏫 {self.name} is teaching {self.subject}")

    def full_details(self):
        self.introduce()              # From Grandparent
        self.show_employee_info()     # From Parent
        self.show_professor_info()    # Own method

# Usage
prof = Professor(
    "Dr. Ananya Singh",
    45,
    "EMP1024",
    150000,
    "Data Structures",
    "Machine Learning"
)
prof.full_details()
print("-" * 30)
prof.teach()
````

````python
# 4️⃣ HIERARCHICAL INHERITANCE -> Multiple child classes inherit from ONE parent class.
# Parent
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def start_engine(self):
        print(f"🔑 Starting {self.brand} {self.model} engine...")

    def show_info(self):
        print(f"Brand: {self.brand}")
        print(f"Model: {self.model}")
        print(f"Year: {self.year}")

# Child 1
class Car(Vehicle):
    def __init__(self, brand, model, year, doors, fuel_type):
        super().__init__(brand, model, year)
        self.doors = doors
        self.fuel_type = fuel_type

    def show_car_info(self):
        self.show_info()
        print(f"Doors: {self.doors}")
        print(f"Fuel Type: {self.fuel_type}")

    def drive(self):
        print(f"🚗 Driving {self.brand} {self.model} on the road!")

# Child 2
class Bike(Vehicle):
    def __init__(self, brand, model, year, engine_cc, has_abs):
        super().__init__(brand, model, year)
        self.engine_cc = engine_cc
        self.has_abs = has_abs

    def show_bike_info(self):
        self.show_info()
        print(f"Engine CC: {self.engine_cc}")
        print(f"ABS: {'Yes' if self.has_abs else 'No'}")

    def ride(self):
        print(f"🏍️ Riding {self.brand} {self.model} on the highway!")

# Child 3
class Truck(Vehicle):
    def __init__(self, brand, model, year, load_capacity, axles):
        super().__init__(brand, model, year)
        self.load_capacity = load_capacity
        self.axles = axles

    def show_truck_info(self):
        self.show_info()
        print(f"Load Capacity: {self.load_capacity} tons")
        print(f"Axles: {self.axles}")

    def haul(self):
        print(f"🚛 Hauling goods with {self.brand} {self.model}!")

# Usage
car = Car("Toyota", "Camry", 2023, 4, "Petrol")
bike = Bike("Royal Enfield", "Classic 350", 2023, 350, True)
truck = Truck("Tata", "Semi", 2022, 25, 3)

print("🚗 CAR:")
car.show_car_info()
car.drive()

print("\n🏍️ BIKE:")
bike.show_bike_info()
bike.ride()

print("\n🚛 TRUCK:")
truck.show_truck_info()
truck.haul()
````

````python
# 5️⃣ HYBRID INHERITANCE  -> A mix of multiple and multilevel inheritance.
# Base classes
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Name: {self.name}, Age: {self.age}")

class Department:
    def __init__(self, dept_name, dept_code):
        self.dept_name = dept_name
        self.dept_code = dept_code

    def show_department(self):
        print(f"Department: {self.dept_name} ({self.dept_code})")

class Grade:
    def __init__(self, grade, cgpa):
        self.grade = grade
        self.cgpa = cgpa

    def show_grade(self):
        print(f"Grade: {self.grade} (CGPA: {self.cgpa})")

# Multiple Inheritance
class Student(Person, Department):
    def __init__(self, name, age, dept_name, dept_code, student_id):
        Person.__init__(self, name, age)
        Department.__init__(self, dept_name, dept_code)
        self.student_id = student_id

    def show_student(self):
        self.introduce()
        self.show_department()
        print(f"Student ID: {self.student_id}")

# Multilevel + Multiple Inheritance
class GradStudent(Student, Grade):
    def __init__(self, name, age, dept_name, dept_code, student_id, grade, cgpa, research_topic):
        Student.__init__(self, name, age, dept_name, dept_code, student_id)
        Grade.__init__(self, grade, cgpa)
        self.research_topic = research_topic

    def show_grad_student(self):
        self.show_student()
        self.show_grade()
        print(f"Research Topic: {self.research_topic}")
        print(f"📚 Thesis: {self.research_topic}")

class UnderGradStudent(Student, Grade):
    def __init__(self, name, age, dept_name, dept_code, student_id, grade, cgpa, club):
        Student.__init__(self, name, age, dept_name, dept_code, student_id)
        Grade.__init__(self, grade, cgpa)
        self.club = club

    def show_undergrad(self):
        self.show_student()
        self.show_grade()
        print(f"Club: {self.club}")
        print(f"🎓 Participating in {self.club} club")

# Usage
grad = GradStudent(
    "Amit Kumar", 25,
    "Data Science", "DS101",
    "S2023G001",
    "A", 9.2,
    "NLP using Deep Learning"
)

undergrad = UnderGradStudent(
    "Sneha Patel", 20,
    "Computer Science", "CS101",
    "S2023U001",
    "B+", 8.5,
    "Coding Club"
)

print("🎓 GRADUATE STUDENT")
print("=" * 40)
grad.show_grad_student()

print("\n" + "=" * 40)
print("🎓 UNDERGRADUATE STUDENT")
print("=" * 40)
undergrad.show_undergrad()
````

````python
# Method Overriding (we need inheritance for this to work)
# ========================================================
# Overriding means: a child class defines a method with the SAME NAME as the parent's method.
# When called on the child object, the child's version WINS — the parent's version is "overridden".


# Parent class — Animal
class Animal:

    r = 23                      # class variable (shared by Animal and its children)

    # Parent's constructor — stores the name on the object
    def __init__(self, name):
        self.name = name

    # Parent's version of details()
    def details(self):
        print(f"Your name is {self.name}")


# Child class — Humans, inherits from Animal
# The "(Animal)" part means: Humans IS-A Animal, so it gets everything Animal has.
class Humans(Animal):

    b = 12                      # class variable (only in Humans, not in Animal)

    # Child's constructor
    # super().__init__(name) calls the PARENT's __init__ so we don't repeat the logic.
    # Without this line, self.name would never be set — and details() would crash.
    def __init__(self, name):
        super().__init__(name)

    # Child's OWN version of details() — this OVERRIDES Animal.details()
    # Same name, same parameters, but different behavior.
    def details(self):
        print(f"Your info is {self.name} and this is all we have")


# Create a Humans object — this calls Humans.__init__, which calls Animal.__init__ via super()
hmn = Humans("Rahul")


# hmn.b → looks in Humans first → found b = 12
print(hmn.b)                    # 12

# hmn.r → not in Humans, so Python walks up to Animal → found r = 23 (inherited)
print(hmn.r)                    # 23

# hmn.details() → Python looks for details() starting at Humans → found it → uses it.
# The Animal version is IGNORED — this is what "overriding" means.
hmn.details()                   # Your info is Rahul and this is all we have
````

````python
# Abstraction: enforce is a rulebook (ABC) saying every vehicle MUST have enginestart(),
# so bike, car, truck each inherit it and give their own behavior.

# Import ABC (Abstract Base Class) and abstractmethod (the rule decorator) from the abc module
from abc import ABC, abstractmethod

# enforce is the RULEBOOK — it cannot be used directly, only inherited.
# Any class that inherits from it MUST implement enginestart().
class enforce(ABC):

    # This is the RULE — abstract means subclasses MUST provide their own version.
    # We only declare the name here; the actual behavior is left to the subclasses.
    @abstractmethod
    def enginestart(self):
        pass  # No body — just a rule that says "this method must exist"


# bike follows the rulebook, so it MUST define enginestart() its own way.
class bike(enforce):
    def enginestart(self):
        print("Bike engine started 🏍️")   # Bike's own way of starting


# car also follows the rulebook, with its own version of enginestart().
class car(enforce):
    def enginestart(self):
        print("Car engine started 🚗")    # Car's own way of starting


# truck also follows the rulebook, with its own version of enginestart().
class truck(enforce):
    def enginestart(self):
        print("Truck engine started 🚚")  # Truck's own way of starting


# Now we actually create real objects from each vehicle class.
Bike  = bike()    # create a bike object
Car   = car()     # create a car object
Truck = truck()   # create a truck object

# Call enginestart() on each object.
# Even though the method name is the SAME, each vehicle runs its OWN version.
Bike.enginestart()    # 🏍️ prints: Bike engine started 🏍️
Car.enginestart()     # 🚗 prints: Car engine started 🚗
Truck.enginestart()   # 🚚 prints: Truck engine started 🚚
````

````python
# Dunder Methods ->Magic Methods
#Dunder methods (short for "Double Under" methods) are special methods in Python whose names begin and end with two underscores (__).
#They are not called directly by you. Python calls them automatically when you use operators, built-in functions, or certain language syntax on your objects.


# Dunder methods are predefined special methods in Python, named with a leading and trailing double underscore (e.g., __init__, __str__, __len__), that allow user-defined classes to integrate with Python's built-in syntax, operators, and functions — enabling behavior like printing, indexing, iteration, comparison, and arithmetic on custom objects.
# Dunder (double underscore) methods let your objects behave like built-in Python types. They're called automatically when you use operators or built-in functions.
````

````python
# Define a class called Animal — it represents an animal with a name.
class Animal:

    # __init__ is a dunder method — Python calls it automatically when you write Animal(...)
    # It takes the animal's name and stores it on the object as self.name.
    def __init__(self, name):
        self.name = name        # save the name into the object


    # __str__ is a dunder method — Python calls it automatically when you use print() or str()
    # Without it, print(obj) would show something ugly like <__main__.Animal object at 0x...>
    # Here we return a friendly sentence that includes the animal's name.
    def __str__(self):
        return (f"Hello my name is {self.name}")   # pretty format with the name


# Create two Animal objects — each call to Animal(...) runs __init__ and stores a different name.
obj   = Animal("Lion")     # obj.name   = "Lion"
obj20 = Animal("Tiger")    # obj20.name = "Tiger"


# print(obj) → Python calls obj.__str__() → prints "Hello my name is Lion"
print(obj)                 # Hello my name is Lion

# print(obj20) → Python calls obj20.__str__() → prints "Hello my name is Tiger"
print(obj20)               # Hello my name is Tiger
````

````python
# Define a class called Vector — it represents a point/arrow in 3D space.
class Vector:

    # __init__ is a dunder method — Python calls it automatically when you write Vector(...)
    # It stores the three coordinates (x, y, z) as attributes on the object.
    def __init__(self, x, y, z):
        self.x, self.y, self.z = x, y, z    # save x, y, z into the object


    # __str__ is a dunder method — Python calls it automatically when you use print() or str()
    # Without it, print(v) would show something ugly like <__main__.Vector object at 0x...>
    def __str__(self):
        return f"({self.x}, {self.y}, {self.z})"   # pretty format: (x, y, z)


    # __add__ is a dunder method — Python calls it automatically when you use the + operator
    # It takes two Vectors, adds them component-by-component, and returns a NEW Vector
    def __add__(self, other):
        return Vector(self.x + other.x,            # new x = x1 + x2
                      self.y + other.y,            # new y = y1 + y2
                      self.z + other.z)            # new z = z1 + z2


    # __len__ is a dunder method — Python calls it automatically when you use len()
    # A 3D vector has 3 components, so we return 3
    def __len__(self):
        return 3


# Create two Vector objects (this calls __init__ each time)
v1 = Vector(1, 2, 3)    # v1 = (1, 2, 3)
v2 = Vector(4, 5, 6)    # v2 = (4, 5, 6)


# print(v1) → Python calls v1.__str__() → prints "(1, 2, 3)"
print(v1)               # (1, 2, 3)

# v1 + v2 → Python calls v1.__add__(v2) → returns a new Vector (5, 7, 9)
# then print() calls __str__ on that new Vector
print(v1 + v2)          # (5, 7, 9)

# len(v1) → Python calls v1.__len__() → returns 3
print(len(v1))          # 3
````

````python
# Decorators
# ========================================================
# A decorator is a function that TAKES another function as input,
# WRAPS it with extra behavior, and RETURNS a new function.
# Think of it as gift wrapping 🎁 — the gift (function) stays the same,
# but the wrapper adds something extra around it.
# The original function's code is NOT modified — we just surround it.


# This is the decorator function itself.
# It takes "fnc" (a function) as its argument.
def extragreeting(fnc):

    # This inner function is the WRAPPER — it's what actually runs
    # whenever the decorated function is called.
    def wrapper():

        # Step 1: extra behavior BEFORE the original function runs
        print("Welcome to team")

        # Step 2: call the ORIGINAL function (the "gift" inside the wrapper)
        fnc()

        # Step 3: extra behavior AFTER the original function runs
        print("Thanks , visit again")

    # Return the wrapper function (NOT the result of calling it)
    return wrapper


# @extragreeting is decorator syntax — it's a shortcut for:
#     greetings = extragreeting(greetings)
# It means: "wrap greetings() inside extragreeting()"
@extragreeting
def greetings():
    print("Good morning")


# Now when we call greetings(), we are actually calling wrapper(),
# which runs the extra lines + the original greetings() + more extra lines.
greetings()
````

````python
# *args and **kwargs
# When you don't know how many arguments a function will receive, use *args (for positional) and **kwargs (for keyword).

# def add(*args):
#   return args

# print(add(23,45,5432,3,45,6,54,32,)) # prints (23, 45, 5432, 3, 45, 6, 54, 32)

def add(*args):
  s=0
  for i in args:
    s+=i
  return s

print(add(23,45,5432,3,45,6,54,32))
````

````python
# *args — Variable Number of Arguments
# ========================================================
# The * before "args" tells Python: "collect all positional arguments
# into a single TUPLE called args".
# So the function can accept ANY number of arguments — 1, 5, 100, whatever.


# Define a function "subs" that accepts any number of arguments.
# Inside the function, "args" will be a tuple containing all the values passed in.
def subs(*args):

    # Start with d = 0 — this will be our running result.
    d = 0

    # Loop through every number that was passed in.
    # "i" takes each value from the tuple one by one.
    for i in args:
        d = d - i      # subtract the current value from d, then store it back

    # Return the final result
    return d


# Call subs() with 8 numbers.
# Python packs them into a tuple: (23, 45, 5432, 3, 45, 6, 54, 32)
# Inside the function, args = (23, 45, 5432, 3, 45, 6, 54, 32)
print(subs(23, 45, 5432, 3, 45, 6, 54, 32))
````

````python
def mul(*args):
  m=1
  for i in args:
    m=m*i
  return m

print(mul(34,67,543,234))
````

````python
# **kwargs — Variable Number of Keyword Arguments
# ========================================================
# The ** before "kwargs" tells Python: "collect all keyword arguments
# (i.e., name=value pairs) into a single DICTIONARY called kwargs".
# So the function can accept ANY number of named arguments — 1, 5, 100, whatever.

# Quick contrast:
#   *args   → collects positional arguments → tuple     → args = (1, 2, 3)
#   **kwargs → collects keyword arguments    → dictionary → kwargs = {'a': 1, 'b': 2}


# Define a function "show" that accepts any number of keyword arguments.
# Inside the function, "kwargs" will be a dict of everything passed in.
def show(**kwargs):

    # Loop through the dictionary — key is the name, value is the value.
    for key, value in kwargs.items():
        print(f"{key} = {value}")


# Call show() with several keyword arguments.
# Python packs them into a dict: {'name': 'Rahul', 'age': 21, 'city': 'Delhi'}
show(name="Rahul", age=21, city="Delhi")
````

````python
# Comprehensions — One-liners
# ========================================================
# A comprehension is a compact way to build a list, dict, or set
# in a SINGLE line — instead of writing a multi-line for-loop + append.
# Syntax pattern:
#     [ expression  for item in iterable  if condition ]
#       ↑ what to    ↑ where items    ↑ optional filter
#         produce      come from


# --------------------------------------------------------
# 1. LIST COMPREHENSION — builds a list
# --------------------------------------------------------

# Basic: square every number from 0 to 4
squares = [x**2 for x in range(5)]
print(squares)                          # [0, 1, 4, 9, 16]

# With filter: keep only even numbers from 0 to 9
evens = [x for x in range(10) if x % 2 == 0]
print(evens)                            # [0, 2, 4, 6, 8]

# With transform + filter: square only the odd numbers from 0 to 9
odd_squares = [x**2 for x in range(10) if x % 2 != 0]
print(odd_squares)                      # [1, 9, 25, 49, 81]

# Uppercase every word in a list
words = ["hello", "world", "python"]
upper_words = [w.upper() for w in words]
print(upper_words)                      # ['HELLO', 'WORLD', 'PYTHON']


# --------------------------------------------------------
# 2. DICT COMPREHENSION — builds a dictionary
# --------------------------------------------------------

# Basic: map each number to its square
squared = {x: x**2 for x in range(5)}
print(squared)                          # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

# From two parallel lists: build a name → age mapping
names = ["Rahul", "Aman", "Priya"]
ages  = [21, 22, 20]
people = {n: a for n, a in zip(names, ages)}
print(people)                           # {'Rahul': 21, 'Aman': 22, 'Priya': 20}

# With filter: keep only ages above 20
adults = {n: a for n, a in people.items() if a > 20}
print(adults)                           # {'Rahul': 21, 'Aman': 22}


# --------------------------------------------------------
# 3. SET COMPREHENSION — builds a set (auto-removes duplicates)
# --------------------------------------------------------

# Remainders when dividing 0..9 by 3 → only unique values kept
unique = {x % 3 for x in range(10)}
print(unique)                           # {0, 1, 2}

# Get unique first letters of words
first_letters = {w[0] for w in words}
print(first_letters)                    # {'h', 'w', 'p'}  (order may vary)


# --------------------------------------------------------
# 4. NESTED COMPREHENSION — comprehension inside comprehension
# --------------------------------------------------------

# Flatten a 2D list into a 1D list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [n for row in matrix for n in row]
print(flat)                             # [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Build a 3x3 multiplication table as a list of lists
table = [[i * j for j in range(1, 4)] for i in range(1, 4)]
print(table)                            # [[1,2,3], [2,4,6], [3,6,9]]


# --------------------------------------------------------
# 5. WITH CONDITIONAL EXPRESSION (if/else inside the output)
# --------------------------------------------------------

# Label each number as "even" or "odd"
labels = ["even" if x % 2 == 0 else "odd" for x in range(5)]
print(labels)                           # ['even', 'odd', 'even', 'odd', 'even']


# --------------------------------------------------------
# 6. GENERATOR EXPRESSION — like a list comp but lazy (uses ( ))
# --------------------------------------------------------

# Doesn't build the list in memory — produces values one at a time
gen = (x**2 for x in range(5))
print(gen)                              # <generator object ...>
print(list(gen))                        # [0, 1, 4, 9, 16]  ← consumed now
# Note: once consumed, the generator is empty
print(list(gen))                        # []
````

````python
# map(), filter(), zip()
# ========================================================
# These are built-in functions that work on iterables (lists, tuples, etc.)
# They are the functional-programming cousins of comprehensions.
# All three return LAZY iterators — wrap them in list() to see the values.


nums = [1, 2, 3, 4, 5]


# --------------------------------------------------------
# map() — TRANSFORM every item
# --------------------------------------------------------
# Applies a function to EVERY item in the iterable and returns the results.
# Syntax: map(function, iterable)
# Here we use a lambda (an anonymous one-line function) to double each number.
# map returns a map object, so we wrap it in list() to see the output.
doubled = list(map(lambda x: x * 2, nums))
print(doubled)                       # [2, 4, 6, 8, 10]

# Equivalent with a comprehension:
# doubled = [x * 2 for x in nums]    # same result, often cleaner


# --------------------------------------------------------
# filter() — KEEP items that pass a test
# --------------------------------------------------------
# Keeps only items where the function returns True.
# The function must return True or False (a boolean test).
# Syntax: filter(function, iterable)
# Here we keep numbers where x % 2 == 0 (i.e., even numbers).
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)                         # [2, 4]

# Equivalent with a comprehension:
# evens = [x for x in nums if x % 2 == 0]   # same result


# --------------------------------------------------------
# zip() — COMBINE two or more lists into pairs
# --------------------------------------------------------
# Pairs up items from multiple iterables, index by index.
# Returns tuples: (item_from_list1, item_from_list2, ...)
# Stops at the SHORTEST list if lengths differ.
names  = ["A", "B", "C"]
scores = [90, 85, 78]

pairs = list(zip(names, scores))
print(pairs)                         # [('A', 90), ('B', 85), ('C', 78)]

# Very common use — build a dict from two parallel lists:
score_dict = dict(zip(names, scores))
print(score_dict)                    # {'A': 90, 'B': 85, 'C': 78}

# Loop through pairs directly — unpack each tuple into two variables:
for name, score in zip(names, scores):
    print(f"{name} scored {score}")
# A scored 90
# B scored 85
# C scored 78
````

````python
# Print a right-angle triangle with stars.
# Take input from the user (convert to int because input() returns a string)
n = int(input("Enter number of rows: "))

# Loop from 1 to n (range(1, n+1) because range stops before the end)
for i in range(1, n + 1):
    # Print "*" repeated i times
    print("*" * i)
````

````python
# Print a mirrored right-angle triangle.
# Take input from the user
n = int(input("Enter number of rows: "))

# Loop from 1 to n
for i in range(1, n + 1):
    # Print (n - i) spaces, then i stars
    print(" " * (n - i) + "*" * i)
````

````python
#Print a centred diamond with stars.
# Take input from the user
n = int(input("Enter an odd number: "))
if n % 2 == 0:
    print("Please enter an odd number.")
else:
      # Top half (including middle row): i = 1 to n
  for i in range(1, n + 1):
      print(" " * (n - i) + "*" * (2 * i - 1))

  # Bottom half: i = n-1 down to 1
  for i in range(n - 1, 0, -1):
      print(" " * (n - i) + "*" * (2 * i - 1))
````
